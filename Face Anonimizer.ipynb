```python
import cv2 
import mediapipe as mp

webcam = cv2.VideoCapture(0)
image_detection = mp.solutions.face_detection 

with image_detection.FaceDetection(model_selection=1 , min_detection_confidence=0.5) as face_detection:
    while True : 
        ret , frame = webcam.read()
        if not ret:
            break
        frame_rgb = cv2.cvtColor(frame , cv2.COLOR_BGR2RGB)
        out = face_detection.process(frame_rgb)
        if out.detections:
            for detection in out.detections:
              location_data = detection.location_data
              bbox = location_data.relative_bounding_box
              xmin = int(bbox.xmin * frame.shape[1])
              ymin = int(bbox.ymin * frame.shape[0])         
              width = int(bbox.width * frame.shape[1])
              height = int(bbox.height * frame.shape[0])      
              frame[ymin : ymin+height , xmin:xmin+width ] = cv2.blur(frame[ymin : ymin+height , xmin:xmin+width]  , (50,50))

            cv2.imshow('Frame' , frame)
            if cv2.waitKey(1) & 0xFF == ord('p'):
              break

webcam.release()
cv2.destroyAllWindows() 
```
