```python
import cv2 
import numpy as np 
import pandas as pd 

def hand_bbox(frame ) : 
    img = cv2.cvtColor(frame , cv2.COLOR_BGR2RGB)
    hand = mp.solutions.hands 
    a = [] 
    b = [] 

    with hand.Hands(max_num_hands = 1 , static_image_mode = False , min_detection_confidence = 0.5 , min_tracking_confidence = 0.5 , model_complexity = 1  ) as mesh : 
        out = mesh.process(img)
        if out.multi_hand_landmarks:
            h,w,_ = img.shape
            for hand_landmarks in out.multi_hand_landmarks :
               for landmarks in hand_landmarks.landmark : 
                   a.append(landmarks.x * h )
                   b.append(landmarks.y * w )
            x_min  = int(min(a))
            x_max = int(max(a))
            y_min = int(min(b))
            y_max = int(max(b))
    return x_min , x_max , y_min , y_max

def bbox_of_objects_covered_from_hands(frame)
    data  = pd.read_csv('data.csv')
    original  = data[['bbox1' , 'bbox2' , 'bbox3' , 'bbox4' ]].values

    possible_detection = []
    x_min , x_max , y_min , y_max = hand_bbox(frame) 
    for box in original :
        for i in range(4) : 
            if box[i] > x_min and box[i] < x_max and box[i] > y_min and box[i] < y_max :
               possible_detection.append(box) 
            possible_detection.append(1.0) 
    possible_detection = np.unique(np.array(possible_detection), axis=0)
    return possible_detection 




```
